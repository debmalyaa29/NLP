```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem import PorterStemmer #it convert every words to its original word
stemming=PorterStemmer()

for word in words:
    print(word +"--->"+stemming.stem(word))
```

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4) #remove the last ing,s,e,able
reg_stemmer.stem('eating')
```

```python
from nltk.stem import SnowballStemmer #it performs better tham PorterStemmer
snowballstemmer=SnowballStemmer('english')
for word in words:
    print(word +"--->"+snowballstemmer.stem(word))
```

```python
print(snowballstemmer.stem("sportingly"))
print(snowballstemmer.stem("fairly"))
```
