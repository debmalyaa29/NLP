```python
paragraph="""
Dr. APJ Abdul Kalam believed that dreams are not what we see while sleeping, but dreams are those that do not let us sleep. He encouraged young people to think big, work hard, and never give up when facing challenges. He believed that education, determination, and a strong vision can help individuals build a better future. According to him, every student has the ability to learn, create, and contribute to the development of the nation. His words continue to inspire millions of young minds to believe in themselves and work toward their goals.
"""
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords # it will remove all the non important words
import nltk
```

```python
stopwords.words('english')
```

```python
stemmer=PorterStemmer()
```

```python
sentences=nltk.sent_tokenize(paragraph)
print(sentences)
```

```python
type(sentences)
```

```python
##Apply stopwords ad filter and then apply stemming
for i in range(len(sentences)):

    words = nltk.word_tokenize(sentences[i])

    words = [
        stemmer.stem(word)
        for word in words
        if word not in set(stopwords.words('english'))
    ]

    sentences[i] = ' '.join(words)#convertinf all the words into sentences  
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer =SnowballStemmer('english')
```

```python
##Apply stopwords ad filter and then apply stemming
for i in range(len(sentences)):

    words = nltk.word_tokenize(sentences[i])

    words = [
       snowballstemmer.stem(word)
        for word in words
        if word not in set(stopwords.words('english'))
    ]

    sentences[i] = ' '.join(words)#convertinf all the words into sentences  
```

```python
sentences
```

```python
#doing with lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
##Apply stopwords ad filter and then apply lemmatization
for i in range(len(sentences)):

    words = nltk.word_tokenize(sentences[i])

    words = [
       lemmatizer.lemmatize(word,pos='v')
        for word in words
        if word not in set(stopwords.words('english'))
    ]

    sentences[i] = ' '.join(words)#convertinf all the words into sentences  
```
